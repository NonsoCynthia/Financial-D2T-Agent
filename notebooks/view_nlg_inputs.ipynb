```python
samples = load_generation_samples(
    dataset_path=str(DATASET_PATH),
    dataset_kind=DATASET_KIND,
    start_date=START_DATE,
    end_date=END_DATE,
    tickers=TICKERS,
    min_stocks_per_month=MIN_STOCKS_PER_MONTH,
)

print(f"Loaded {len(samples)} samples.")
```

```python
catalog = pd.DataFrame(
    [
        {
            "sample_id": sample["sample_id"],
            "sample_name": sample.get("sample_name"),
            "analysis_date": sample.get("analysis_date"),
            "ticker_count": sample.get("ticker_count"),
            "tickers": ", ".join(sample.get("tickers", [])),
            "previous_analysis_date": sample.get("previous_analysis_date"),
        }
        for sample in samples
    ]
)

catalog
```

```python
SELECTED_SAMPLE_ID = 1
SELECTED_ANALYSIS_DATE = None  # example: "2025-03-31"


def get_sample(sample_list, sample_id=None, analysis_date=None):
    if analysis_date:
        for sample in sample_list:
            if str(sample.get("analysis_date", ""))[:10] == analysis_date:
                return sample
        raise KeyError(f"No sample found for analysis_date={analysis_date!r}")

    for sample in sample_list:
        if int(sample["sample_id"]) == int(sample_id):
            return sample
    raise KeyError(f"No sample found for sample_id={sample_id!r}")


sample = get_sample(
    samples,
    sample_id=SELECTED_SAMPLE_ID,
    analysis_date=SELECTED_ANALYSIS_DATE,
)

sample["sample_name"], sample["analysis_date"], sample.get("tickers")
```

```python
sample_summary = {
    "sample_id": sample.get("sample_id"),
    "sample_name": sample.get("sample_name"),
    "sample_type": sample.get("sample_type"),
    "analysis_date": sample.get("analysis_date"),
    "tickers": sample.get("tickers"),
    "ticker_count": sample.get("ticker_count"),
    "previous_analysis_date": sample.get("previous_analysis_date"),
    "previous_report_preview": str(sample.get("previous_report", "N/A"))[:300],
}

pd.Series(sample_summary)
```

```python
facts_df = pd.DataFrame(sample["data_input"], columns=["subject", "relation", "object"])
facts_df
```

```python
stocks_df = pd.DataFrame(
    [
        {
            "ticker": row["ticker"],
            "recommendation": row["recommendation"],
            "target_price": row["target_price"],
            "recommendation_justification": row["recommendation_justification"],
            "indicator_count": len(row.get("indicators", {})),
        }
        for row in sample.get("stocks", [])
    ]
)
stocks_df
```

```python
SELECTED_TICKER = sample["stocks"][0]["ticker"] if sample.get("stocks") else None

if SELECTED_TICKER is None:
    print("No stock rows available in this sample.")
else:
    stock_row = next(row for row in sample["stocks"] if row["ticker"] == SELECTED_TICKER)
    display(pd.Series({
        "ticker": stock_row["ticker"],
        "recommendation": stock_row["recommendation"],
        "target_price": stock_row["target_price"],
        "recommendation_justification": stock_row["recommendation_justification"],
        "indicator_count": len(stock_row.get("indicators", {})),
    }))

    indicators_df = pd.DataFrame(
        [
            {"indicator": key, "value": value}
            for key, value in sorted(stock_row.get("indicators", {}).items())
        ]
    )
    display(indicators_df)
```

```python
selected_month = str(sample.get("analysis_date", ""))[:10]
month_outputs = sorted(DATASET_PATH.rglob(f"*_{selected_month}_output_*.json"))
month_manager = sorted(DATASET_PATH.rglob(f"*_{selected_month}_manager_decision_*.json"))

max_len = max(len(month_outputs), len(month_manager), 1)
artifact_table = pd.DataFrame(
    {
        "output_files": [
            str(month_outputs[i].relative_to(PROJECT_ROOT)) if i < len(month_outputs) else None
            for i in range(max_len)
        ],
        "manager_decision_files": [
            str(month_manager[i].relative_to(PROJECT_ROOT)) if i < len(month_manager) else None
            for i in range(max_len)
        ],
    }
)
artifact_table
```
